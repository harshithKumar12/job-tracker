# Agent Job Tracker

This notebook runs the same offline-safe pipeline as the CLI. It prints drafts only; it never sends mail.

```python
from agent_job_tracker.config import Config
from agent_job_tracker.email_fetcher import fetch_threads
from agent_job_tracker.pipeline import print_digest, process_threads
from agent_job_tracker.sample_data import load_sample_data

config = Config.from_env()
if config.has_any_imap_credentials and not config.has_imap_credentials:
    raise RuntimeError('Set both IMAP_USERNAME and IMAP_PASSWORD, or leave both unset for offline mode.')
threads = fetch_threads(config) if config.has_imap_credentials else load_sample_data()
print_digest(process_threads(threads, config))
```
